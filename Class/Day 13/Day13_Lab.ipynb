### Model Building

```python
y = df['medv']
```

```python
x = df.iloc[:,:13]
```

```python
x
```

```python
sns.displot(df.medv)
```

```python
sns.boxplot(df.medv)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=2)
```

```python
sns.distplot(y_train)
```

```python
x_train
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
model_1 = lr.fit(x_train,y_train)
```

```python
model_1.intercept_
```

```python
model_1.coef_
```

```python
x_test[:1]
```

```python
model_1.predict(x_test[:1])
```

```python
pred = model_1.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error
```

```python
mean_squared_error(y_test,pred)
```

```python
mean_absolute_error(y_test,pred)
```

```python
err = pd.DataFrame({'Actual':y_test,'Pred':pred})
```

```python
err
```

```python
np.sqrt(mean_squared_error(y_test,pred))
```

```python
err.reset_index(drop=True,inplace=True)
```

```python
plt.plot(err)
plt.legend(['Actual','Predicted'])
plt.show()
```

### Decision Trees

```python
from sklearn.tree import DecisionTreeRegressor
from dtreeviz import model
```

```python
dt = DecisionTreeRegressor(max_depth=5)
```

```python
model_2 = dt.fit(x_train,y_train)
```

```python
tree_viz = model(model_2,x_train,y_train,target_name='medv',feature_names=x.columns)
```

```python
tree_viz.view()
```
