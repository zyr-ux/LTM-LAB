```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import History
from matplotlib import pyplot as plt
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('../Day 14/churn.csv')
df.head()
```

```python
x = df.iloc[:,2:-1]
y = df.iloc[:,-1]
```

```python
x = pd.get_dummies(x,columns=['Geography','Gender'],drop_first=True)
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2,random_state=77)
```

```python
scaler = StandardScaler()
xtrain = scaler.fit_transform(xtrain)
xtest = scaler.fit_transform(xtest)
```

```python
model = Sequential()
```

```python
model.add(Dense(units=6,activation='relu',kernel_initializer='uniform',input_shape=(11,)))
#model.add(Dropout(rate=0.1))
model.add(Dense(units=3,activation='relu',kernel_initializer='uniform'))
model.add(Dense(units=1,activation='sigmoid',kernel_initializer='uniform'))
```

```python
model.compile(optimizer='rmsprop',loss='binary_crossentropy',metrics=['precision','accuracy'])
```

```python
from keras.callbacks import EarlyStopping
early_stop = EarlyStopping(monitor='precision',patience=20,min_delta=0.001,restore_best_weights=True,verbose=1)
```

```python
model.fit(xtrain,ytrain,batch_size=25,epochs=200,validation_split=0.2,callbacks=[early_stop])
```
