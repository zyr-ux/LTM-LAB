```python
# Ask the user for number of students. for each students ask for marks of 3 subjects and then calculate average marks of each student and print "qualified" if avg marks > 75 otherwise "disqualified"
```

```python
s_num = int(input("Enter the number of students"))

for i in range(s_num):
    s1=int(input("Enter Maths marks"))
    s2=int(input("Enter Physics marks"))
    s3=int(input("Enter Chemistry marks"))
    avg=(s1+s2+s3)/3
    if(avg>75):
        print("qualified")
    else:
        print("disqualified")
```

```python
# write a fn to find out whether a given number is prime or not

def prime(x):
    c=0
    for i in range(2,x):
        if(x%i==0):
            c+=1
    return((True if c==0 else False))

prime(7)
```

```python
# Use the fucntion to print the primes from 2 to 10

for i in range(2,10):
    if prime(i) == True:
        print(f"{i} is a prime number")
    else:
        print(f"{i} is not a prime number")
```
