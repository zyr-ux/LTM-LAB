```python
# Duck Typing Polymorphism

class car:
    def move(self):
        return 'drive on the road'
    
class boat:
    def move(self):
        return 'cruise in the water'

```

```python
def start_journey(vehicle):
    print(vehicle.move())

```

```python
start_journey(car())
```

```python
start_journey(boat())
```

```python
class organization:
    org_name='LTM'
    def __init__(self,name,num_emp,city):
        self.name=name
        self.num_emp=num_emp
        self.city=city

    def intro(self):
        print('Welcome to the organization')

    def check(self):
        print(f'{self.city}')



class school(organization):
    def __init__(self,name,num_emp,num_students,city):
        super().__init__(name,num_emp,city)
        self.num_students=num_students
```

```python
xaviers = school("St. Xaviers Institution", 108888, 5000, 'Panihati')
```

```python
xaviers.check()
```

```python
class book:
    def __init__(self,title,author,pages):
        self.title=title
        self.author=author
        self.pages=pages

    def __str__(self):
        return f'The {self.title} is written by {self.author} and has {self.pages} pages'

    def __len__(self):
        return self.pages

    def __del__(self):
        print(f"The book {self.title} is deleted")
```

```python
try:
    x=100
    y = int(input('enter a number'))
    print(x/y)
except ArithmeticError as A:
    print(A)
except TypeError as T:
    print(T.args)
except ValueError as V:
    print(V)
else:
    print("Successfully worked")
finally:
    print("this runs irrespective of try except")
```
