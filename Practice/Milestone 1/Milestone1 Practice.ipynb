```python
def create_orders_array(orders_data: list) -> np.ndarray:
    return np.array(orders_data, dtype=np.int64)

def validate_orders_array(orders_array: np.ndarray) -> bool:
    if len(orders_array) == 0:
        return False
    
    for item in orders_array:
        if not isinstance(item, (int,float)):
            return False

    for item in orders_array:
        if item < 0:
            return False

    return True

def compute_order_metrics(orders_array:np.ndarray) -> tuple:
    total = int(np.sum(orders_array))
    average = round(np.mean(orders_array),2)
    max = int(np.max(orders_array))

    return (total,average,max)

def categorize_order_levels(orders_array:np.ndarray) -> np.ndarray:
    labels = []

    for volume in orders_array:
        if volume < 300:
            labels.append('Low Volume')
        elif volume < 500:
            labels.append('Moderate Volume')
        else:
            labels.append('High Volume')

    return np.array(labels)

def longest_order_growth_streak(orders_array:np.ndarray) -> int:
    max_streak = -1
    current_streak = -1

    for i,value in enumerate(orders_array):
        if i>0 and value > orders_array[i-1]:
            current_streak+=1
        else:
            current_streak = 1

        max_streak = max(max_streak, current_streak)

    return max_streak

def format_order_counts(orders_array:np.ndarray) -> np.ndarray:
    formatted_orders = []

    for order in orders_array:
        formatted_orders = format(order)

    return np.array(formatted_orders)
```

### Question 1.4

```python
class DeliveryTimeAnalyzer:

    def create_delivery_log_df(self, delivery_log:list) -> pd.DataFrame:
        return pd.DataFrame(delivery_log, columns = ['OrderID','RestaurantCode','Area','DeliveryTime'])

    def create_restaurant_master_df(self, restaurant_master:list) -> pd.DataFrame:
        return pd.DataFrame(restaurant_master, columns=['RestaurantCode','RestaurantName'])

    def merge_restaurant_names(self, delivery_log_df: pd.DataFrame, 
                            restaurant_master_df: pd.DataFrame) -> pd.DataFrame:
        return pd.merge(restaurant_master_df,delivery_log_df,'left',on='RestaurantCode')

    def average_delivery_by_restaurant(self,merged_df:pd.DataFrame) -> pd.DataFrame:
        return merged_df.groupby('RestaurantName')['DeliveryTime'].mean().reset_index().rename(columns = {'DeliveryTime':'Average Delivery'})

    def filter_slow_deliveries(self,delivery_log_df:pd.DataFrame, threshold:int) -> pd.DataFrame:
        return delivery_log_df[delivery_log_df['DeliveryTime']>threshold]

    def slowest_delivery_area(self, delivery_log_df:pd.DataFrame) -> pd.DataFrame:
        avg_delivery_times = delivery_log_df.groupby('Area')['DeliveryTime'].mean().reset_index().rename(columns = {'DeliveryTime':'Average Delivery'})
        max_delay_value = avg_delivery_times['Average Delivery'].max()
        return avg_delivery_times[avg_delivery_times['Average Delivery'] == max_delay_value].head(1).reset_index(drop=True)
```

### Question 1.3

```python
class MovieRatingAnalyzer:


    def create_rating_array(self, ratings_array: list) -> np.ndarray:
        return np.array(ratings_array,dtype=int)

    def validate_ratings(self,rating_array:np.ndarray) -> bool:
        if rating_array.shape == 0:
            return False

        return bool(np.all((rating_array >= 0) & (rating_array<=100)))

    def compute_rating_summary(self, rating_array:np.ndarray) -> tuple:
        numeric_array = np.array(rating_array,float)
        total = np.sum(numeric_array)
        average = np.mean(numeric_array)
        max = np.max(numeric_array)

        return (total,average,max)

    def apply_bonus(self,ratings_array) -> np.ndarray:
        float_arr = np.array(ratings_array,dtype=float)

        bonus_arr = np.where(float_arr > 85, round(float_arr*1.05,1), float_arr)

        return np.clip(bonus_arr,None,100)

    def categorize_movies(self,rating_array:np.ndarray) ->np.ndarray:

        rating_result = []

        for value in rating_array:
            if value >= 90:
                rating_result.append('Excellent')
            elif value >=80:
                rating_result.append('Good')
            else:
                rating_result.append('Needs Improvement')

        return np.array(rating_result,dtype=str)

    def format_ratings_with_grades(self, rating_array:np.ndarray) -> np.ndarray:
        grade = []

        for item in rating_array:
            if item >= 90:
                grade.append('A')
            elif item >= 80:
                grade.append('B')
            elif item >= 70:
                grade.append('C')
            else:
                grade.append('D')

        return np.array(grade,dtype=str)
```
