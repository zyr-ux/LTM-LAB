```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import classification_report,accuracy_score,precision_score,recall_score

import joblib
```

```python
df = pd.read_csv('hospital.csv')
df.head()
```

```python
df.isna().sum()
```

```python
x = df.iloc[:,:-1]
x.head()
```

```python
y = df.iloc[:,-1]
y.head()
```

```python
x = x.drop(['patient_id'],axis=1)
x
```

```python
x = pd.get_dummies(x,columns=['insurance_type'],drop_first=True)
x['gender'] = x['gender'].map({'Male':0,'Female':1})
x
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2,random_state=77)
```

```python
lr = LogisticRegression(solver='newton-cholesky')
```

```python
model_lr = lr.fit(xtrain,ytrain)
```

```python
pred_lr = model_lr.predict(xtest)
```

```python
report = classification_report(ytest,pred_lr)
print(report)
```

```python
joblib.dump(model_lr,'hospital_model.joblib')
```
