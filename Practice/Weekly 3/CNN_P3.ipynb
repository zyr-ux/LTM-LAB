# CNN Practice 3

```python
import keras
from keras import layers
from sklearn.metrics import accuracy_score, precision_score
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
(xtrain, ytrain), (xtest, ytest) = keras.datasets.fashion_mnist.load_data()
```

```python
xtrain = xtrain / 255.0
xtest = xtest / 255.0
```

```python
# 1. One-hot encode the labels (shape changes from (60000,) to (60000, 10))
ytrain = keras.utils.to_categorical(ytrain, num_classes=10)
ytest = keras.utils.to_categorical(ytest, num_classes=10)
```

```python
model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPool2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPool2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(256, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(
    xtrain,
    ytrain,
    batch_size=50,
    epochs=5
)
```

```python
test_loss, test_accuracy = model.evaluate(xtest, ytest)
test_accuracy
```

```python
pred = model.predict(xtest)
pred = pred.argmax(axis=1)

# Convert one-hot ytest back to 1D class indices for sklearn metrics
ytest_labels = ytest.argmax(axis=1)

acc = accuracy_score(ytest_labels, pred)
prec = precision_score(ytest_labels, pred, average='weighted')
print(f'Accuracy = {acc}\nPrecision = {prec}')
```
