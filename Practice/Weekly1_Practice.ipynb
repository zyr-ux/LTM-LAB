### Q1

```python
def get_even_numbers(numbers):
    result=[]
    for number in numbers:
        if number%2==0:
            result.append(number)

    return result
```

```python
get_even_numbers([7,12,9,18,21,24])
```

### Q3

```python
def valid_prices(prices):
    result = []
    for price in prices:
        if price>0:
            result.append(price)

    return result
```

```python
valid_prices([250,-10,500,0,125])
```

### Q4

```python
def category_count(categories):
    count = {}
    for category in categories:
        count[category] = count.get(category,0) + 1

    return count
```

```python
category_count(['Electronics','Books','Books','Clothing','Electronics'])
```

### Q6

```python
def rank_employees(score, limit):
    result = []

    result = sorted(score.items(),key=get_value,reverse=True)
    return result[:limit]

def get_value(item):
    return item[1]
```

```python
rank_employees({'Amit':82,'Neha':95,'Ravi':76,'Priya':90}, 2)
```
